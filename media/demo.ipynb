# Notebook Cell Id — point your coding agent at an exact cell

Every cell since nbformat 4.5 carries a stable `id` — VS Code just never shows it.
This extension puts it in each cell's status bar (bottom right → `⧉ #<index> <id>`, click to copy),
so instead of pasting a cell into the agent chat, or saying "the third code cell", you write:

> **“Review cell e66c99c5”** &nbsp;·&nbsp; **“Fix cell 8d3f2a1c”**

- **Accurate** — the badge shows the same `id` the agent sees in the `.ipynb` JSON, and ids
  survive re-ordering, inserting, and deleting cells. Indices don't.
- **Token-efficient** — 8 characters instead of the cell's whole contents.

```python
def moving_average(xs, window=3):
    return [sum(xs[i:i + window]) / window for i in range(len(xs))]

moving_average([1, 2, 3, 4, 5])  # tail is wrong → "Fix cell e66c99c5"
```

```python
# "cell 2" changes meaning the moment someone inserts a cell above.
print("Move this cell anywhere — it stays 8d3f2a1c")
```
